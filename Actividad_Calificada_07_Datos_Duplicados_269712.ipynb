### 1.1 Conteo de duplicados exactos (5 pts)

Calcula el número de filas marcadas como duplicadas en `hoteles` considerando **todas** las columnas, con `keep='first'`, `keep='last'` y `keep=False`. Imprime los tres resultados con un texto que explique qué significa cada uno.

```python
# Escribe tu código aquí

df_duplicados_first = hoteles.duplicated(keep='first')
print('Duplicados keep first:Representa la duplicidad de datos marcando primera coincidencia como False.(podria entenderse como un escaneo de arriba hacia abajo)\n',df_duplicados_first.sum(),' Filas duplicadas\n')

df_duplicados_last = hoteles.duplicated(keep='last')
print('Duplicados keep last:Represena la duplicidad de datos, marcando la primera coincidencia como True. (podria entenderse como un escaneo de abajo hacia arriba)\n',df_duplicados_last.sum(),' Filas duplicadas\n')

df_duplicados_false = hoteles.duplicated(keep=False)
print('Duplicados keep False:Marca todos los datos duplicados sin excepcion como True, y deja en False solo los que no estan duplicados.\n',df_duplicados_false.sum(),' Filas duplicadas\n')
```

### 1.2 Magnitud del problema (5 pts)

Con los resultados de 1.1, calcula e imprime:

- El **porcentaje** de filas sobrantes respecto al total.
- El número de **grupos repetidos** (combinaciones de valores que aparecen más de una vez).

```python
# Escribe tu código aquí


# 1. Cálculo del porcentaje de filas sobrantes respecto al total
total_filas = len(hoteles)
filas_sobrantes = df_duplicados_first.sum()
porcentaje_sobrantes = (filas_sobrantes / total_filas) * 100

# 2. Cálculo del número de grupos repetidos (combinaciones únicas que aparecen más de una vez)
# Usamos value_counts() para contar las ocurrencias de cada combinación idéntica de filas
conteo_combinaciones = hoteles.value_counts(dropna=False)
numero_grupos_repetidos = (conteo_combinaciones > 1).sum()

# Impresión de resultados con formato claro
print(f"Porcentaje de filas sobrantes: {porcentaje_sobrantes:.2f}%")
print(f"Número de grupos repetidos   : {numero_grupos_repetidos}")
```

### 1.3 Duplicados por hotel (5 pts)

Determina **cuántas** filas sobrantes (`keep='first'`) hay en cada hotel y qué **porcentaje** representan respecto al total de reservas de ese hotel.

Sugerencia: crea una copia de `hoteles`, agrega una columna booleana `es_duplicado` con el resultado de `duplicated()` y agrupa por `hotel`.

```python
# Escribe tu código aquí

# Creamos una copia del DataFrame y agregamos una columna booleana de duplicados (keep='first')
hoteles_dup = hoteles.copy()
hoteles_dup['es_duplicado'] = hoteles_dup.duplicated(keep='first')

# Agrupamos por hotel para contar reservas totales y filas sobrantes
resumen_hotel = hoteles_dup.groupby('hotel').agg(
    total_reservas=('is_canceled', 'size'),
    filas_sobrantes=('es_duplicado', 'sum')
)

# Calculamos el porcentaje de duplicados respecto al total de cada hotel
resumen_hotel['porcentaje'] = (resumen_hotel['filas_sobrantes'] / resumen_hotel['total_reservas']) * 100

print(resumen_hotel)
```

## Actividad 3. Impacto en el análisis (25 puntos)

### 3.1 Versión sin duplicados (5 pts)

Crea `hoteles_limpio` eliminando los duplicados exactos, conservando la primera aparición y **renumerando el índice**. Imprime el número de filas antes y después y cuántas se eliminaron.

```python
# Escribe tu código aquí

# Creamos hoteles_limpio eliminando duplicados exactos, conservando la primera y renumerando índice
hoteles_limpio = hoteles.drop_duplicates(keep='first', ignore_index=True)

print(f"Filas antes     : {len(hoteles)}")
print(f"Filas después   : {len(hoteles_limpio)}")
print(f"Filas eliminadas: {len(hoteles) - len(hoteles_limpio)}")
```

### 3.2 Métricas por hotel (8 pts)

Para `hoteles` y para `hoteles_limpio`, calcula por hotel con `groupby()`:

- Número de reservas (`.size()`).
- Tasa de cancelación (media de `is_canceled`).
- Tarifa diaria promedio (media de `adr`).

Nota: para calcular la media de varias columnas a la vez, selecciona una lista de columnas: `df.groupby('hotel')[['col1', 'col2']].mean()`.

Guarda las medias en dos `DataFrame`, `resumen_original` y `resumen_limpio`, con `hotel` como columna (usa `reset_index()`).

```python
# Escribe tu código aquí

def calcular_metricas(df):
    # Calcula reservas, tasa de cancelación y adr medio agrupado por hotel
    return df.groupby('hotel').agg(
        reservas=('is_canceled', 'size'),
        tasa_cancelacion=('is_canceled', 'mean'),
        adr_medio=('adr', 'mean')
    ).reset_index()

resumen_original = calcular_metricas(hoteles)
resumen_limpio = calcular_metricas(hoteles_limpio)

print("Resumen Original:")
print(resumen_original)
print("\nResumen Limpio:")
print(resumen_limpio)
```

### 3.3 Tabla comparativa (6 pts)

Agrega a cada resumen una columna `version` (`'original'` y `'sin duplicados'`) y únelos con `pd.concat()` en un `DataFrame` llamado `comparacion`, con índice renumerado y ordenado por `hotel`, de modo que las dos versiones de cada hotel queden juntas.

```python
# Escribe tu código aquí

# Agregamos la columna 'version' a cada resumen
resumen_original['version'] = 'original'
resumen_limpio['version'] = 'sin duplicados'

# Unimos ambos DataFrames con pd.concat y ordenamos por hotel
comparacion = pd.concat([resumen_original, resumen_limpio], ignore_index=True).sort_values(by='hotel').reset_index(drop=True)
print(comparacion)
```
