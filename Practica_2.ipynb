```python
import re

ruta_archivo = '/content/drive/MyDrive/Colab Notebooks/Programación/Semana 1/Practica 2/informe_acciones.txt'

with open(ruta_archivo, 'r', encoding='latin-1') as archivo:
    contenido = archivo.read()

# Captura el código en mayúsculas (ej. CSCO) y el valor
patron_ticker_valor = r'Informe de Acci[oó]n - ([A-Z0-9\.]+).*?Valor actual:\s*\$([0-9,]+\.?[0-9]*)'

coincidencias = re.findall(patron_ticker_valor, contenido, re.DOTALL)

print("==================================================")
for ticker, valor in coincidencias:
    print(f"Empresa: {ticker:<6} ---> Valor de accion: ${valor}")
print("==================================================")
```
