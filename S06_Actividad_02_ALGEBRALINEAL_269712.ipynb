## Parte 1 — Vectores: perfiles de producto

Una tienda en línea representa cada producto como un vector de 3 componentes: calificación de calidad, calificación de precio, y calificación de servicio (escala 1 a 5, donde 5 es mejor). Ejecuta la siguiente celda para cargar los tres productos — **no la modifiques**.

````python
producto_A = np.array([4.5, 3.2, 5.0])
producto_B = np.array([4.0, 2.8, 4.7])
producto_C = np.array([1.0, 5.0, 1.2])

print(producto_A, producto_B, producto_C)
````

---

### Ejercicio 1 — Norma de los vectores (10 pts)

Calcula la norma de `producto_A`, `producto_B` y `producto_C` usando `np.linalg.norm()`.

````python
# Tu código aquí

# Calculamos la norma (magnitud) de cada vector con np.linalg.norm()
norma_A = np.linalg.norm(producto_A)
norma_B = np.linalg.norm(producto_B)
norma_C = np.linalg.norm(producto_C)

# Imprimimos los resultados
print(f"Norma Producto A: {norma_A:.4f}")
print(f"Norma Producto B: {norma_B:.4f}")
print(f"Norma Producto C: {norma_C:.4f}")

````

---

### Ejercicio 2 — Producto punto y ángulo (15 pts)

1. Calcula el producto punto entre `producto_A` y `producto_B`, y entre `producto_A` y `producto_C`, usando `np.dot()` o `@`.
2. Calcula el ángulo (en grados) entre `producto_A` y `producto_B`, y entre `producto_A` y `producto_C`.
3. Con base en los ángulos obtenidos, indica (en un comentario o celda de texto) cuál de los dos productos (`B` o `C`) es más similar a `producto_A`, y por qué el ángulo es el indicador correcto para esa comparación.

````python
# Tu código aquí


# 1. Producto punto usando np.dot() o @
dot_AB = np.dot(producto_A, producto_B)
dot_AC = np.dot(producto_A, producto_C)

# 2. Coseno del ángulo y conversión a grados con np.arccos() y np.degrees()
cos_AB = dot_AB / (norma_A * norma_B)
angulo_AB_deg = np.degrees(np.arccos(cos_AB))

cos_AC = dot_AC / (norma_A * norma_C)
angulo_AC_deg = np.degrees(np.arccos(cos_AC))

print(f"Producto punto A y B: {dot_AB:.4f} | Ángulo: {angulo_AB_deg:.2f}°")
print(f"Producto punto A y C: {dot_AC:.4f} | Ángulo: {angulo_AC_deg:.2f}°")
````

El Producto B es más similar al Producto A, ya que el ángulo entre sus vectores es de tan solo 3.03°, en comparación con los 65.54° entre A y C.
El ángulo es el indicador correcto porque mide la orientación relativa de los vectores en el espacio (la proporción relativa entre calidad, precio y servicio), ignorando la magnitud o escala absoluta del vector. Dos productos con proporciones similares de atributos apuntarán en una dirección casi identica.

### Ejercicio 3 — Normalización (10 pts)

Normaliza los tres vectores (`producto_A`, `producto_B`, `producto_C`) para obtener sus vectores unitarios. Verifica, imprimiendo el resultado, que la norma de cada vector normalizado sea 1.

````python
# Tu código aquí

# Obtenemos los vectores unitarios dividiendo cada vector entre su propia norma
unitario_A = producto_A / norma_A
unitario_B = producto_B / norma_B
unitario_C = producto_C / norma_C

# Verificamos e imprimimos que la norma de cada vector unitario sea exactamente 1.0
print("Unitario A:", np.round(unitario_A, 4), "| Norma:", np.linalg.norm(unitario_A))
print("Unitario B:", np.round(unitario_B, 4), "| Norma:", np.linalg.norm(unitario_B))
print("Unitario C:", np.round(unitario_C, 4), "| Norma:", np.linalg.norm(unitario_C))

````

### Ejercicio 7 — Síntesis: distancia y similitud (20 pts)

Retomando los tres productos de la Parte 1:

1. Calcula la **distancia euclidiana** entre `producto_A` y `producto_B`, y entre `producto_A` y `producto_C`, usando la norma de la diferencia entre los vectores (`np.linalg.norm(a - b)`).
2. Calcula la **similitud coseno** entre los mismos pares — es el cociente que ya calculaste en el Ejercicio 2 antes de aplicar `arccos()`: $\frac{a \cdot b}{\|a\| \|b\|}$.
3. Con base en ambas métricas (distancia y similitud coseno), concluye en una cadena de texto impresa cuál producto (`B` o `C`) es más parecido a `producto_A`, y si ambas métricas coinciden en la misma conclusión.

````python
# Tu código aquí

# 1. Distancia euclidiana usando la norma de la diferencia: np.linalg.norm(a - b)
dist_AB = np.linalg.norm(producto_A - producto_B)
dist_AC = np.linalg.norm(producto_A - producto_C)

# 2. Similitud coseno: (a . b) / (||a|| * ||b||)
sim_cos_AB = np.dot(producto_A, producto_B) / (norma_A * norma_B)
sim_cos_AC = np.dot(producto_A, producto_C) / (norma_A * norma_C)

# 3. Conclusión en cadena de texto impresa
resumen_conclusion = f"""
=== EVALUACIÓN DE SIMILITUD ENTRE PRODUCTOS ===

1. Distancia Euclidiana:
   - Distancia (A, B): {dist_AB:.4f}
   - Distancia (A, C): {dist_AC:.4f}

2. Similitud Coseno:
   - Similitud Coseno (A, B): {sim_cos_AB:.4f}
   - Similitud Coseno (A, C): {sim_cos_AC:.4f}

CONCLUSIÓN FINAL:
El Producto B es significativamente más parecido al Producto A que el Producto C.
Ambas métricas coinciden plenamente en esta conclusión:
- El Producto B presenta una menor distancia euclidiana ({dist_AB:.4f} < {dist_AC:.4f}).
- El Producto B presenta una mayor similitud coseno ({sim_cos_AB:.4f} > {sim_cos_AC:.4f}).
"""

print(resumen_conclusion)
````
